#### Ejercicio 20: Decoradores para Idempotencia


```python
# CONTEXTO: Para no ensuciar la lógica de negocio de tus funciones con validaciones de red,
# usas un decorador de Python que "envuelve" la función. Este intercepta el ID de la petición
# y decide si le da paso al código real o cancela la ejecución duplicada.
def asegurar_idempotencia(funcion_original):
    historial_peticiones = set()

    def envoltura(request_id, *args, **kwargs):
        # Si el ID de red ya está dentro del conjunto (set) de control
        if request_id ____________ historial_peticiones:
            print("Petición clonada detectada por red. Operación bloqueada con seguridad.")
            return None
        historial_peticiones.add(request_id)
        return funcion_original(request_id, *args, **kwargs)
    return envoltura
```
